```python
# DNA Sequence Analysis

# Step 1: Input DNA
dna = input("Enter DNA sequence: ").upper()

print("DNA Sequence:", dna)
print("Length:", len(dna))

# Step 2: GC Content
gc = dna.count('G') + dna.count('C')
print("GC Content:", round(gc / len(dna) * 100, 2), "%")

# Step 3: Find Motif
motif = input("Enter motif: ").upper()
pos = dna.find(motif)

if pos != -1:
    print("Motif Position:", pos + 1)
else:
    print("Motif Not Found")

# Step 4: Reverse Complement
comp = dna.translate(str.maketrans("ATGC", "TACG"))[::-1]
print("Reverse Complement:", comp)

# Step 5: Find Coding Region
start = dna.find("ATG")
stop = dna.find("TGA", start + 3)

if start != -1 and stop != -1:
    print("Coding Region:", dna[start:stop + 3])
else:
    print("Coding Region: Not Found")

# Step 6: Analysis Report
print("\nDNA Analysis Report")
print("DNA Length:", len(dna))
print("GC Content:", round(gc / len(dna) * 100, 2), "%")
```
